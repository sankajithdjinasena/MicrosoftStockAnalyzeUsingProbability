### Import Neceessary Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from scipy.stats import norm, lognorm, t, skew, kurtosis
import warnings
warnings.filterwarnings('ignore')

# Set style
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")
```

### Data Loading

```python
df = pd.read_csv('microsoft_stock.csv')
df.head()
```

```python
df.columns = df.columns.str.lower().str.replace(' ', '_')
df['date'] = pd.to_datetime(df['date'])
df = df.sort_values('date')
df.set_index('date', inplace=True)
```

### Feature Engineering

```python
df['daily_return'] = df['close'].pct_change()
df.head()
```

```python
df['log_return'] = np.log(df['close'] / df['close'].shift(1))
df.head()
```

```python
df['volatility_30d'] = df['daily_return'].rolling(window=30).std()
df.head()
```

```python
df["volatility_30d"].unique()
```

```python
df['price_change'] = df['close'].diff()
df.head()
```

```python
df_clean = df.dropna()
df_clean.shape
```

### **Descriptive Statistical & Probability Measures**

```python
returns = df_clean['daily_return']
log_returns = df_clean['log_return']
```

```python
stats_dict = {
    'Mean Return': returns.mean(), # Average return value
    'Median Return': returns.median(), # Median return value
    'Std Deviation': returns.std(), # Standard deviation of returns
    'Variance': returns.var(), # Variance of returns
    'Skewness': skew(returns), # Skewness of returns
    'Kurtosis': kurtosis(returns),  # Kurtosis of returns
    'Min Return': returns.min(), # Minimum return value
    'Max Return': returns.max(), # Maximum return value
    'Range': returns.max() - returns.min() # Range of returns
}

for key, value in stats_dict.items():
    print(f"{key:.<50} {value:.6f}")

    
```

### **Distribution Analysis**

Fit Normal Distribution

```python
mu, sigma = norm.fit(returns) 
print(f"Normal Distribution Parameters:")
print(f"  Mean (μ): {mu:.6f}")
print(f"  Std Dev (σ): {sigma:.6f}")
```

Fit Log-Normal Distribution


```python
shape, loc, scale = lognorm.fit(df_clean['close'], floc=0)
print(f"\nLog-Normal Distribution Parameters:")
print(f"  Shape: {shape:.6f}")
print(f"  Scale: {scale:.6f}")
```

Fit t-distribution


```python
df_t, loc_t, scale_t = t.fit(returns)
print(f"\nStudent's t-Distribution Parameters:")
print(f"  Degrees of Freedom: {df_t:.2f}")
print(f"  Location: {loc_t:.6f}")
print(f"  Scale: {scale_t:.6f}")
```

### **Probability Calculation**

```python
# Probability of positive returns
prob_positive = (returns > 0).mean()
f"P(Return > 0): {prob_positive:.4f} ({prob_positive*100:.2f}%)"
```

```python
# Probability of negative returns
prob_negative = (returns < 0).mean()
f"P(Return < 0): {prob_negative:.4f} ({prob_negative*100:.2f}%)"
```

```python
# Probability of returns > 1%
prob_large_positive = (returns > 0.01).mean()
print(f"P(Return > 1%): {prob_large_positive:.4f} ({prob_large_positive*100:.2f}%)")
```

```python
# Probability of returns < -1%
prob_large_negative = (returns < -0.01).mean()
print(f"P(Return < -1%): {prob_large_negative:.4f} ({prob_large_negative*100:.2f}%)")
```

```python
# Probability within 1 std dev
prob_1std = ((returns >= mu - sigma) & (returns <= mu + sigma)).mean()
print(f"P(μ - σ ≤ Return ≤ μ + σ): {prob_1std:.4f} ({prob_1std*100:.2f}%)")
```

```python
# Probability within 2 std dev
prob_2std = ((returns >= mu - 2*sigma) & (returns <= mu + 2*sigma)).mean()
print(f"P(μ - 2σ ≤ Return ≤ μ + 2σ): {prob_2std:.4f} ({prob_2std*100:.2f}%)")
```

#### **Probability Calculation**

```python
# Create shifted returns for conditional analysis
df_clean['prev_return'] = df_clean['daily_return'].shift(1)
df_cond = df_clean.dropna()
```

```python
# P(positive today | positive yesterday)
pos_after_pos = df_cond[(df_cond['prev_return'] > 0) & (df_cond['daily_return'] > 0)].shape[0]
pos_yesterday = df_cond[df_cond['prev_return'] > 0].shape[0]
prob_pos_given_pos = pos_after_pos / pos_yesterday if pos_yesterday > 0 else 0
print(f"P(Positive Today | Positive Yesterday): {prob_pos_given_pos:.4f}")
```

```python
# P(negative today | positive yesterday)
neg_after_pos = df_cond[(df_cond['prev_return'] > 0) & (df_cond['daily_return'] < 0)].shape[0]
prob_neg_given_pos = neg_after_pos / pos_yesterday if pos_yesterday > 0 else 0
print(f"P(Negative Today | Positive Yesterday): {prob_neg_given_pos:.4f}")
```

```python
# P(positive today | negative yesterday)
pos_after_neg = df_cond[(df_cond['prev_return'] < 0) & (df_cond['daily_return'] > 0)].shape[0]
neg_yesterday = df_cond[df_cond['prev_return'] < 0].shape[0]
prob_pos_given_neg = pos_after_neg / neg_yesterday if neg_yesterday > 0 else 0
print(f"P(Positive Today | Negative Yesterday): {prob_pos_given_neg:.4f}")
```

```python
# P(negative today | negative yesterday)
neg_after_neg = df_cond[(df_cond['prev_return'] < 0) & (df_cond['daily_return'] < 0)].shape[0]
prob_neg_given_neg = neg_after_neg / neg_yesterday if neg_yesterday > 0 else 0
print(f"P(Negative Today | Negative Yesterday): {prob_neg_given_neg:.4f}")
```

#### **Bayes' Theorem Application**

```python
volatility_threshold = df_clean['volatility_30d'].quantile(0.75)
df_clean['high_volatility'] = df_clean['volatility_30d'] > volatility_threshold
df_bayes = df_clean.dropna(subset=['high_volatility'])
```

```python
# P(Large Move)
large_move = abs(df_bayes['daily_return']) > 0.02
p_large_move = large_move.mean()
```

```python
# P(High Vol)
p_high_vol = df_bayes['high_volatility'].mean()
```

```python
# P(Large Move | High Vol)
high_vol_mask = df_bayes['high_volatility']
p_large_given_high_vol = large_move[high_vol_mask].mean()
```

```python
p_high_vol_given_large = (p_large_given_high_vol * p_high_vol) / p_large_move if p_large_move > 0 else 0
```

```python
print(f"P(Large Move > 2%): {p_large_move:.4f}")
print(f"P(High Volatility): {p_high_vol:.4f}")
print(f"P(Large Move | High Volatility): {p_large_given_high_vol:.4f}")
print(f"P(High Volatility | Large Move) [Bayes]: {p_high_vol_given_large:.4f}")
```

#### **Expected value & Variance**

```python
expected_return = returns.mean()
print(f"Expected Daily Return (E[X]): {expected_return:.6f}")
```

```python
# Expected annual return (assuming 252 trading days)
expected_annual_return = (1 + expected_return)**252 - 1
print(f"Expected Annual Return: {expected_annual_return:.4f} ({expected_annual_return*100:.2f}%)")
```

```python
# Variance and volatility
variance = returns.var()
volatility = returns.std()
print(f"Variance (Var[X]): {variance:.8f}")
print(f"Daily Volatility (σ): {volatility:.6f}")
```

```python
# Annual volatility
annual_volatility = volatility * np.sqrt(252)
print(f"Annual Volatility: {annual_volatility:.4f} ({annual_volatility*100:.2f}%)")
```

```python
# Value at Risk (VaR) at 95% confidence
var_95 = np.percentile(returns, 5)
print(f"Value at Risk (95% confidence): {var_95:.6f}")
```

```python
# Conditional Value at Risk (CVaR/Expected Shortfall)
cvar_95 = returns[returns <= var_95].mean()
print(f"Conditional VaR (Expected Shortfall): {cvar_95:.6f}")
```

```python
# Sharpe Ratio (assuming risk-free rate of 2%)
risk_free_rate = 0.02 / 252  # Daily risk-free rate
sharpe_ratio = (expected_return - risk_free_rate) / volatility if volatility > 0 else 0
annual_sharpe = sharpe_ratio * np.sqrt(252)
print(f"Sharpe Ratio (Annual): {annual_sharpe:.4f}")
```

#### **Markov Chain Analysis**

```python
# Define states: Up, Down, Flat
def classify_state(ret):
    if ret > 0.001:
        return 'Up'
    elif ret < -0.001:
        return 'Down'
    else:
        return 'Flat'

df_clean['state'] = df_clean['daily_return'].apply(classify_state)
df_clean['next_state'] = df_clean['state'].shift(-1)

# Create transition matrix
states = ['Up', 'Down', 'Flat']
transition_matrix = pd.DataFrame(0, index=states, columns=states, dtype=float)

for current_state in states:
    current_mask = df_clean['state'] == current_state
    total_current = current_mask.sum()
    
    if total_current > 0:
        for next_state in states:
            count = ((df_clean['state'] == current_state) & 
                    (df_clean['next_state'] == next_state)).sum()
            transition_matrix.loc[current_state, next_state] = count / total_current

print("\nTransition Probability Matrix:")
print(transition_matrix)
```

#### **Monte Carlo Simulation**

```python
# Simulate future prices
np.random.seed(42)
n_simulations = 1000
n_days = 252  # 1 year

last_price = df_clean['close'].iloc[-1]
simulation_df = pd.DataFrame()

for i in range(n_simulations):
    prices = [last_price]
    for _ in range(n_days):
        # Generate random return from fitted normal distribution
        random_return = np.random.normal(mu, sigma)
        new_price = prices[-1] * (1 + random_return)
        prices.append(new_price)
    simulation_df[i] = prices

# Calculate probabilities
final_prices = simulation_df.iloc[-1]
prob_above_current = (final_prices > last_price).mean()
prob_double = (final_prices > last_price * 2).mean()
prob_loss = (final_prices < last_price * 0.9).mean()

print(f"Current Price: ${last_price:.2f}")
print(f"Simulations: {n_simulations}, Days: {n_days}")
print(f"P(Price > Current after 1 year): {prob_above_current:.4f}")
print(f"P(Price doubles): {prob_double:.4f}")
print(f"P(Price drops >10%): {prob_loss:.4f}")
print(f"Expected Price (mean): ${final_prices.mean():.2f}")
print(f"Median Price: ${final_prices.median():.2f}")
print(f"95% Confidence Interval: [${final_prices.quantile(0.025):.2f}, ${final_prices.quantile(0.975):.2f}]")
```
